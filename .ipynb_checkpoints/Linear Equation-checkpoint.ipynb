````python
## *Solve Linear Equations Recursively*

**Explanation:**
Equations with multiple unknowns (variables) can be solved step by step. One common way is to 
**eliminate a variable** from some equations until we are left with a smaller set of equations. 
Eventually, we reduce it to a single equation with one variable, solve it, and then substitute 
back to find the remaining variables.

For example, suppose we have two equations:

1. `2x + y = 5`
2. `x - y = 1`


We can **eliminate** one variable, say `y`, to solve for `x`. Then we use the value of `x` to find `y`.

We will use two helper functions:

* `eliminate_variable(equations, var_index)`: removes one variable from the system of equations.
* `solve_for_first_variable(equations)`: directly solves when there is only one variable left.

Using **recursion**, we keep reducing the problem until only one variable remains.

---

**Exercise:**
Write a function `solve_equations(equations)` that:

* Takes `equations` as a list of lists, where each sublist represents one linear equation in the
form `[a1, a2, ..., an, b]`, meaning:

  $$
  a_1x_1 + a_2x_2 + \dots + a_nx_n = b
  $$
* Uses `eliminate_variable()` and `solve_for_first_variable()` to recursively solve for all variables.
* Returns the list of variable values.

---

**Example:**

```python
# Example system:
# 2x + y = 5
# x - y = 1
equations = [
    [2, 1, 5],  # 2x + y = 5
    [1, -1, 1]  # x - y = 1
]

solve_equations(equations)  
# Expected Output: [2.0, 1.0]
# Meaning: x = 2, y = 1
```

```python
# Another example:
# x + y + z = 6
# 2y + 5z = -4
# 2x + 5y - z = 27
equations = [
    [1, 1, 1, 6],    # x + y + z = 6
    [0, 2, 5, -4],   # 2y + 5z = -4
    [2, 5, -1, 27]   # 2x + 5y - z = 27
]

solve_equations(equations)  
# Expected Output: [5.0, 3.0, -2.0]
# Meaning: x = 5, y = 3, z = -2
```

---
````

````python
def multiply_polynomial(poly, num):
    new_poly = []
    for p in poly:
        val = num * p
        new_poly.append(val)
    return new_poly 


def add_polynomials(p1, p2):
    rev_p1 = p1[::-1]
    rev_p2 = p2[::-1]
    add_poly = []
    
    len_p1 = len(p1)
    len_p2 = len(p2)
    
    if( len_p1 == len_p2):
        
        for i in range(len(p1)):
            sum_val = rev_p1[i] + rev_p2[i]
            add_poly.append(sum_val)
            
    elif len_p1 < len_p2:
        for i in range(len(p1)):
            sum_val = rev_p1[i] + rev_p2[i]
            add_poly.append(sum_val)
            
        add_poly.extend(rev_p2[len_p1:])   #adding the extra elements of p2 as p2 has more elements 
        
    else:
        for i in range(len(p2)):
            sum_val = rev_p1[i] + rev_p2[i]
            add_poly.append(sum_val)
            
        add_poly.extend(rev_p1[len_p2:])   #adding the extra elements of p1 as p1 has more elements 
        
            
    return add_poly[::-1]

def eliminate_variable(eq, var_index):
    coefficient_1 = eq[0][var_index]
    coefficient_2 = eq[1][var_index]
    
    final_eq1 = multiply_polynomial(eq[0],coefficient_2)
    final_eq2 = multiply_polynomial(eq[1],coefficient_1 * -1)
    
    eliminated_eq = add_polynomials(final_eq1,final_eq2)
    
    if(eliminated_eq[var_index] == 0):
#         eliminated_eq = eliminated_eq[0:var_index] + eliminated_eq[var_index:]
#         del eliminated_eq[var_index]
        return eliminated_eq
    else:
        return "some error"


    
def solve_for_first_variable(equation, vars):
    print("equation is ",equation)
    constant_val = equation[-1]
    left_sum = 0
#     eq_index = 1
    vars_index = 0
    for e in equation[1:len(equation)-1]:
        print("e is ", e, "vars is ",vars[vars_index])
        left_sum = left_sum + (e * vars[vars_index])
        vars_index = vars_index + 1
    print("left_sum ", left_sum)    
    output = (constant_val - left_sum ) /  equation[0]
    return output 

def eliminate(eq, index):
    coeficient = eq[index]
    return_eq = []
    for i in range(len(eq)):
        if i == index:
            return_eq.append(0)
#         elif i == len(eq)-1:
#             return_eq.append(eq[i]/coeficient)
        else:
            return_eq.append(-1 * eq[i]/coeficient)
    print("return_eq is ", return_eq) 
    return return_eq
````

````python
equations = [
    [1, 1, 1, 6],    # x + y + z = 6
    [2, -1, 1, 3],   # 2y + 5z = -4
    [1, 2, -1, 4]   # 2x + 5y - z = 27
]



def solve_equations(equations):
    
    
#     variable = [0,-1,-1,6]
    variable = eliminate(equations[0],0)
    eq = equations[1:]
    for i in range(len(eq)):
        new_eq = multiply_polynomial(variable, eq[i][0])
        eq[i][0] = 0
        eq[i] = add_polynomials(eq[i], new_eq)
    
    print(eq)
    
    variable1 = eliminate(eq[0],1)
    print("variable1 ",variable1)
    eq1 = eq[1:]
    print("eq1 is ",eq1)
    for i in range(len(eq1)):
        new_eq1 = multiply_polynomial(variable1, eq1[i][1])
        print("new_eq1 is ",new_eq1)
        eq1[i][1] = 0
        eq1[i] = add_polynomials(eq1[i], new_eq1)
    
    print(eq1)
    
    z = eq1[0][-1]/eq1[0][-2]
    print ("z is ", z)
    
   
    y = solve_for_first_variable(eq[0][1:],[z])
    print("y is",y)
    
    x = solve_for_first_variable(equations[0],[y,z])
    print("x is ",x)
    
````

````python
def multiply_polynomial(poly, num):
    new_poly = []
    for p in poly:
        val = num * p
        new_poly.append(val)
    return new_poly

def add_polynomials(p1, p2):
    rev_p1 = p1[::-1]
    rev_p2 = p2[::-1]
    add_poly = []
    
    len_p1 = len(p1)
    len_p2 = len(p2)
    
    if( len_p1 == len_p2):
        
        for i in range(len(p1)):
            sum_val = rev_p1[i] + rev_p2[i]
            add_poly.append(sum_val)
            
    elif len_p1 < len_p2:
        for i in range(len(p1)):
            sum_val = rev_p1[i] + rev_p2[i]
            add_poly.append(sum_val)
            
        add_poly.extend(rev_p2[len_p1:])   #adding the extra elements of p2 as p2 has more elements 
        
    else:
        for i in range(len(p2)):
            sum_val = rev_p1[i] + rev_p2[i]
            add_poly.append(sum_val)
            
        add_poly.extend(rev_p1[len_p2:])   #adding the extra elements of p1 as p1 has more elements 
        
            
    return add_poly[::-1]


def solve_for_first_variable(equation, vars):
    print("equation is ",equation, " vars are ",vars)
    constant_val = equation[-1]
    left_sum = 0
#     eq_index = 1
    vars_index = 0
    for e in equation[1:len(equation)-1]:
        print("e is ", e, "vars is ",vars[vars_index])
        left_sum = left_sum + (e * vars[vars_index])
        vars_index = vars_index + 1
    print("left_sum ", left_sum)    
    output = (constant_val - left_sum ) /  equation[0]
    return output 
````

````python
def _solve_equations_final(equations):
    
    if len(equations) == 1:
        val = equations[0][1] / equations[0][0]
        return [val] # return nth variable value
    
    # reduce 1 equation and eliminate first variable (0 position)
    first_eq = equations[0] #Picks first equation
    eq = equations[1:]
    eliminated_eq = []
    
    for i in range(len(eq)):
        eq1_multiplier = first_eq[0]
        new_eq1 = multiply_polynomial(eq[i], eq1_multiplier)
        
        eq2_multiplier = -1 * eq[i][0]
        new_eq2 = multiply_polynomial(first_eq,eq2_multiplier)
        
        final_eq = add_polynomials(new_eq1, new_eq2)
        final_eq = final_eq[1:]
        eliminated_eq.append(final_eq)
    print("eliminited eq are ",eliminated_eq)
    result = solve_equations_final(eliminated_eq)
    
    variable_val = solve_for_first_variable(first_eq,result)
    result.append(variable_val)
    return result    
        
def solve_equations_final(equations):
    result = _solve_equations_final(equations)
    return result[::-1]
    
    
    
````

````python
equations = [
    [1, 1, 1, 6],    # x + y + z = 6
    [2, -1, 1, 3],   # 2x -y + z = 3
    [1, 2, -1, 4]   # x + 2y - z = 4
]

solve_equations_final(equations)
````
