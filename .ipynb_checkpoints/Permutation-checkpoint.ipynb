```python
def permute(val):
    
    if len(val) == 1:
        return [val]
    
    final_result = []

    char = val[:1]
    remaining_val = val[1:]
    print(f"character is : {char} and remaining val is : {remaining_val}")
    
    result = permute(remaining_val)
    print("result is ",result)
    
    for r in result:
        
        for i in range(len(r)+1):
            new_val = r[:i] + char + r[i:]
            final_result.append(new_val)
            
            
    return final_result
        
        
        
    
```

```python
def permute1(val):
    
    if len(val) == 1:
        return [val]
    
    final_result = []
    
    for i in range(len(val)):
        char = val[i]
        remaining_val = val[:i] + val[i+1:]
        print(f"character is : {char} and remaining val is : {remaining_val}")

        result = permute1(remaining_val)
        print("result is ",result)

        for r in result:
            new_val = char+r 
            final_result.append(new_val)


    return final_result
```

```python
permute1('abc')
```

```python
def permute_backtracking(input_val,final_value,length):
    
    if(len(final_value) == length):
        print(final_value)
        return
    
    for i in range(len(input_val)):
        new_final_value = input_val[i] + final_value
        remaining_val = input_val[:i] + input_val[i+1:]
        permute_backtracking(remaining_val,new_final_value,length)
        
    
```

```python
permute_backtracking("abcd","",4)
```

```python
def permute_backtracking1(input_val,final_value,length,result):
    
    if(len(final_value) == length):
        result.append(final_value)
        return
    
    for i in range(len(input_val)):
        new_final_value = input_val[i] + final_value
        remaining_val = input_val[:i] + input_val[i+1:]
        permute_backtracking1(remaining_val,new_final_value,length,result)
```

```python
# nested loop  approch

s1 = 'abcde'
s2 = ''
s3 = ''
s4 = ''
s = []
for letter1 in s1:
    s2 = s1.replace(letter1, '')
    #print(s2)
    for letter2 in s2:
        #print(s2)
        s3 = s2.replace(letter2, '')
        for letter3 in s3:
            #print(s3)
            s4 = s3.replace(letter3, '')
            for letter4 in s4:
                #print(s3)
                s5 = s4.replace(letter3, '')
                print(letter1 + letter2 + letter3 + letter4 + letter5)
```

```python
# converting nested loop approach to recursion


# s = "abcd"

def print_perms(s):
    result = []
    _print_perms([], s, result)
    return result
    
def _print_perms(letters, s, result):
    print(f"----- _print_perms ({letters}, {s})", )
    if len(s) == 0:
        r = ''.join(letters)
        result.append(r)
        print("P:" + r)
        return
    for letter in s:
        s1 = s.replace(letter, '')
        print(f'letter:{letter}, s1:{s1}, s:{s}')
        letters.append(letter)
        _print_perms(letters, s1, result)
        letters.pop()
```

```python
print_perms("abc")
```

```python
# Breath First Search
def pploop(s):
    tasks = [([], s)] # database
    while len(tasks) > 0:
        print('current tasks: ', tasks)
        letters, s1 = tasks.pop()
        if len(s1) == 0:
            print("P:", letters)
        for letter1 in s1:
            s2 = s1.replace(letter1, '')
            letters1 = letters.copy()
            letters1.append(letter1)
            new_task = letters1, s2
            tasks.append(new_task)
```

```python
pploop('abc')
```
