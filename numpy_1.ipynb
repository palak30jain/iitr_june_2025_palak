```python
# image is the 3d object we have to convert it into 2d grayscale.


img_gray = []

for row in img:
    for col in row:
        new_col = []
        sum_color = 0
        for color in col:
            sum_color += color
        avg_color =  sum_color / 3
        new_col.append(avg_color)
    img_gray.append(new_col)     
```

```python
plt.imshow(img_gray, cmap= 'gray')
```
