#For rotating the image by 270 degree

* we can do by changing rows into columns and columns into rows
* befor that we create the array of zeros using the rows and columns (size) od original image

```python
row, col, ch = img.shape

img1 = np.zeros((col, row, ch))

# it is giving zero in decimal so the image is nor rendering properly
```

```python
# here in np.zeros we specify that we want integers
img2 = np.zeros((col, row, ch), dtype= 'uint8')
```

```python
img1
```

```python
img.shape
```

```python
img1.shape
```

```python
for i in range(row):
    for j in range(col):
        img1[j][i] = img[i][j]
```

```python
plt.imshow(img1)
```

```python
for i in range(row):
    for j in range(col):
        img2[j][i] = img[i][j]
```

```python
plt.imshow(img2)

# this is 270 degree rotation
```

```python
plt.imshow(img2[: , :: -1])

# reverse the column it is 90 degree rotation
```
