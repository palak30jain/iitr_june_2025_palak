```python
class Node:
    def __init__(self, val, left=None, right=None):
        # print("constructor called")
        self.data = val
        self.L = left
        self.R = right

    def printNode(self):
        # print("printing node")
        if self.L is not None:
            self.L.printNode()
        print(self.data)
        if self.R is not None:
            self.R.printNode()

    def min_node(self):
#         left_min = float('inf')
#         right_min = float('inf')
        middle_min = 0
        left_min = 0
        right_min = 0

        # Get minimum from left subtree
        if self.L is not None:
            left_min = self.L.min_node()

        # Current node's value
        middle_min = self.data

        # Get minimum from right subtree
        if self.R is not None:
            right_min = self.R.min_node()

        return min(left_min, middle_min, right_min)

    def max_node(self):
#         left_max = float('-inf')
#         right_max = float('-inf')
        
        middle_max = 0
        left_max = 0
        right_max = 0

        # Get maximum from left subtree
        if self.L is not None:
            left_max = self.L.max_node()

        # Current node's value
        middle_max = self.data

        # Get maximum from right subtree
        if self.R is not None:
            right_max = self.R.max_node()

        return max(left_max, middle_max, right_max)


# Construct the binary tree
root = Node(2)
root.L = Node(4)
root.R = Node(0)
root.L.L = Node(8)
root.L.R = Node(5)
root.R.L = Node(-13)
root.R.R = Node(70)


```

```python
# Uncomment to print nodes in order
root.printNode()

```

```python
# Print min and max node values
print("Minimum node value:", root.min_node())
print("Maximum node value:", root.max_node())
```

```python
class DecisionNode:
    
    def __init__(self, boundary,boundary_val, left = None, right = None):
        self.boundary = boundary
        self.boundary_val = boundary_val
        self.left = left
        self.right = right
        
     
    def check(self, job_description):
        
        if self.boundary in job_description:
            job_description_value = job_description[self.boundary]
            
            if job_description_value <= self.boundary_val:
                self.left.check(job_description)
            else:
                self.right.check(job_description)
            
        else:
            print(f"{self.boundary} not present in job description")
            
    def preety_print(self, indent = ""): 
        
        print(f"{indent}if {self.boundary} <= {self.boundary_val} then")
        self.left.preety_print(indent = indent + "\t")
        
        print(f"{indent}if {self.boundary} > {self.boundary_val} then")
        self.right.preety_print(indent = indent + "\t")
        
    def DecisionNode_to_dict(self):
        final_dict = {}
        final_dict["boundary"] = self.boundary
        final_dict["boundary_val"] = self.boundary_val
        final_dict["left"] = self.left.DecisionNode_to_dict()
        final_dict["right"] = self.right.DecisionNode_to_dict()
        
        return final_dict
        
        
        
        
```

```python
class Yes(DecisionNode):
    def __init__(self):
        print("Yes constructor")
        
    def check(self, job_description):
        print("Yes")
        
    def preety_print(self, indent = ""):
        print(f"{indent} Yes")
        
    def DecisionNode_to_dict(self):
        return "Yes"

        
class No(DecisionNode):
    def __init__(self):
        print("No constructor")
        
    def check(self, job_description):
        print("No")
        
    def preety_print(self, indent = ""):
        print(f"{indent} No") 
        
    def DecisionNode_to_dict(self):
        return "Yes"    
```

```python
_left = DecisionNode(
    "distance",
    50,
    DecisionNode(
        "coffee",
        1,
        No(),
        DecisionNode(
            "nightshift",
            1,
            Yes(),
            No()
        )
    ),
    No()

)

_right = DecisionNode(
    "distance",
    60,
    DecisionNode(
        "nightshift",
        1,
        Yes(),
        No()
    
    ),
    No()

)




decision = DecisionNode(
    "salary",
    1000,
    _left,
    _right
)
```

```python
job_description = {
    "salary": 15000,
    "distance": 60,
    "coffe": 1,
    "nightshift":0
}


decision.check(job_description)
```
