```python
# ["+", 1, 2] # 6
# ["-", 2, 3] # -1
# ["*", 1, 2] # 2
# ["/", 1, 2] # .5
# ["^", 2, 5] # 32 
# ["+", 1, ["*", 1, 2]] # ["+", 1, 2] == 2

def calculator_2_digit(arg):
    
    exp = arg[0]
    par1 = 0
    par2 = 0
    if type(arg[1]) == list:
        par1 = calculator_2_digit(arg[1])
    else:
        par1 = arg[1]
    
    if type(arg[2]) == list:
        par2 = calculator_2_digit(arg[2])
    else:
        par2 = arg[2]
        
        
    if(exp == '+'):
        return par1 + par2
    elif(exp == '-'):
        return par1 - par2
    elif (exp == '/'):
        return par1 / par2
    elif (exp == '*'):
        return par1 * par2
    
    
# this code is not taking care of [1], [-1]   
# it is taking expression as number and return error (index out of bound) 
```
