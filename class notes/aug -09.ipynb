```python
def multiply(a,b):
    result = 0
    for i in range(a):
        result += b
        print(i)
    return result

```

```python
def multiply_opt(a,b):
    result = 0
    
    if (b < a ):
        a,b = b,a #swaping as we wanrt to loop over minimum to reduce the complexity
    for i in range(a):
        result += b
        print(i)
    return result
```

```python
multiply(4,3)
```

```python
multiply_opt(4,3)
```

```python
def multiply_rec(a,b):
    result = 0
    
    if(b == 0):
        print(result)
        return result
    result = a + multiply_rec(a,b-1)
    print("result is ",result)
    return result
    
```

```python
def divide_rec(divident, divisor):
    
#     rem = divident - divisor
    if ((divident - divisor) < divisor):
        return 1 , (divident - divisor)
    quotient , reminder = divide_rec(divident - divisor,divisor)
    
    return quotient + 1 , reminder
```

```python
divide_rec(20,6)
```

```python
def divide_rec1(divident, divisor):
    
#     rem = divident - divisor
    if ((divident - divisor) < 0): # if divident < divisor:
        return 0 , divident
    quotient , reminder = divide_rec1(divident - divisor,divisor)
    
    return quotient + 1 , reminder
```
