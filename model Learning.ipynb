```python
class TwoPointLine:
    
    def __init__(self):
        self.m = 0
        self.c = 0
    
    def _fit(self,c):
        x1,y1 = c[0]
        x2,y2 = c[1]
        
        if x1 == x2: # verticAL LINE
            return 0,0

        self.m = (y2 - y1) / (x2 - x1)
        self.c = y1 - (self.m * x1)
        
    
    
    
    def fit(self,X,Y):
        coordinates = []
        if(len(X) != len(Y)):
            print("The value of X and Y are not proper")
        
        for i in range(len(X)):
            x_coordinate = X[i][0]
            y_coordinate = Y[i]

            coordinates.append((x_coordinate, y_coordinate))
        #predict the value of m and c using 2 coordinates

        self._fit(coordinates)

    def _predict(self,x):
        
        return (self.m * x) + self.c
    
    
    def predict(self,x_list):
        
        y_val = []
        
        for x in x_list:
            val = self._predict(x[0])
            y_val.append(val)
            
        return y_val 
        




    
    
```

```python
def multiply_polynomial(poly, num):
    new_poly = []
    for p in poly:
        val = num * p
        new_poly.append(val)
    return new_poly

def add_polynomials(p1, p2):
    rev_p1 = p1[::-1]
    rev_p2 = p2[::-1]
    add_poly = []
    
    len_p1 = len(p1)
    len_p2 = len(p2)
    
    if( len_p1 == len_p2):
        
        for i in range(len(p1)):
            sum_val = rev_p1[i] + rev_p2[i]
            add_poly.append(sum_val)
            
    elif len_p1 < len_p2:
        for i in range(len(p1)):
            sum_val = rev_p1[i] + rev_p2[i]
            add_poly.append(sum_val)
            
        add_poly.extend(rev_p2[len_p1:])   #adding the extra elements of p2 as p2 has more elements 
        
    else:
        for i in range(len(p2)):
            sum_val = rev_p1[i] + rev_p2[i]
            add_poly.append(sum_val)
            
        add_poly.extend(rev_p1[len_p2:])   #adding the extra elements of p1 as p1 has more elements 
        
            
    return add_poly[::-1]


def solve_for_first_variable(equation, vars):
    print("equation is ",equation, " vars are ",vars)
    constant_val = equation[-1]
    left_sum = 0
#     eq_index = 1
    vars_index = 0
    for e in equation[1:len(equation)-1]:
        print("e is ", e, "vars is ",vars[vars_index])
        left_sum = left_sum + (e * vars[vars_index])
        vars_index = vars_index + 1
    print("left_sum ", left_sum)    
    output = (constant_val - left_sum ) /  equation[0]
    return output 
```

```python
def _solve_equations_final(equations):
    
    if len(equations) == 1:
        val = equations[0][1] / equations[0][0]
        return [val] # return nth variable value
    
    # reduce 1 equation and eliminate first variable (0 position)
    first_eq = equations[0] #Picks first equation
    eq = equations[1:]
    eliminated_eq = []
    
    for i in range(len(eq)):
        eq1_multiplier = first_eq[0]
        new_eq1 = multiply_polynomial(eq[i], eq1_multiplier)
        
        eq2_multiplier = -1 * eq[i][0]
        new_eq2 = multiply_polynomial(first_eq,eq2_multiplier)
        
        final_eq = add_polynomials(new_eq1, new_eq2)
        final_eq = final_eq[1:]
        eliminated_eq.append(final_eq)
    print("eliminited eq are ",eliminated_eq)
    result = _solve_equations_final(eliminated_eq)
    
    variable_val = solve_for_first_variable(first_eq,result)
    result.append(variable_val)
    return result    
        
def solve_equations_final(equations):
    result = _solve_equations_final(equations)
    return result[::-1]
    
    
    
```

```python
def distance(A,B):
    l= zip(A,B)
    final_list = list(l)
    sq_sum = 0
    for e in final_list:
        sq_sum += ( e[0] -e[1])**2
    d = sq_sum ** 0.5
    return d
        

def find_nearest_neighbour(points, target):
    near_dis = distance(points[0],target)
    near_index  = 0
    
    for index, p in enumerate(points[1:]):
        
        dis = distance(p,target)
        if (dis < near_dis):
            print(f"new near is {p} , new near dis is {dis}, old near dis is {near_dis}")
            near_dis = dis
            near_index = index
            
    return near_index 
```

```python
class NN:
    
    def __init__(self):
        self.points = []
        self.output = []
    
    
    def fit(self,X,Y):   
        self.points = X
        self.output = Y

    def predict(self,p):
        
        index = find_nearest_neighbour(self.points, p)
        return self.output[index]
        
         
```

```python
def binary_search(lst, n):
    start = 0
    end = len(lst)-1
    
    
    while start <= end:
        
        mid = (start + end) // 2
        print("loop", mid)
        if( n < lst[mid]): # left side
            end = mid - 1
        elif n > lst[mid]: # right side
            start = mid + 1
        else:
            return mid
    print("outside loop")
    return -1
            
    
```

```python
binary_search([1, 3, 5, 7, 8, 10, 15],10)
```

```python
def binary_search_recursive(lst, n, start, end):
    if start > end:
        return -1
    
    mid = (start + end) //2
    
    if  n < lst[mid]: #left side
        return binary_search_recursive(lst, n , start, mid-1)
    elif n > lst[mid]: #right side
        return binary_search_recursive(lst, n , mid + 1, end)
    else:
        return mid
```
