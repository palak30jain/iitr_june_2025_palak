```python
def total(A):
    
    if len(A) == 1:
        return A[0]
    return A[0] + total(A[1:])
```

```python
total([1,2,3,4,5,6])
```

```python
def min_2(a,b):
    
    if a<b:
        return a
    return b
    
def myMin(A):
    if len(A) == 1:
        return A[0]
    
    A_1 = myMin(A[1:])
    return min_2(A[0],A_1) 
```

```python
myMin([1,2,3,4,-15,-6,-10])
```

```python
def max_2(a,b):
    
    if a<b:
        return b
    return a
    
def myMax(A):
    if len(A) == 1:
        return A[0]
    
    A_1 = myMax(A[1:])
    return max_2(A[0],A_1) 
```

```python
myMax([1,2,3,4,-15,-6,-10,12])
```

```python
# generalise version



def my_reduce(A, f):
    if len(A) == 1:
        return A[0]
    
    A_1 = my_reduce(A[1:],f)
    return f(A[0],A_1)
```

```python
my_reduce([1,2,3,4,-15,-6,-10,12],max_2)
```

```python
my_reduce([1,2,3,4,-15,-6,-10,12],min_2)
```

```python
# use reduce function to calculate reduce and count in single short


A = [(2,1),(3,1),(5,1),(15,1)]
def sum_count(f, s):
    sm = f[0] + s[0]
    cnt = f[1] + s[1]
    return sm, cnt

my_reduce(A, sum_count)
```

```python
# reduce function can be starte from end as well

def myreduce_end(A, f):
    if len(A) == 1:
        return A[-1]
    else:
        A_1 = myreduce(A[:-1], f)
        return f(A[-1], A_1)
```

```python
# using map we can do operation on each element of array and it will return new array

def mymap(A, f):
    new_arr = []
    for e in A:
        new_el = f(e)
        new_arr.append(new_el)
    return new_arr
```

```python
A = [1,2,3,4]
# double all elements
def dbl(x):
    return x*2
mymap(A, dbl)
```

```python
A = [1,2,3, 4]

def to_tuple(x):
    return (x , 1)
A = mymap(A, to_tuple)
print("After map: ", A)

def add_sm(f, s):
    sm = f[0] + s[0]
    total = f[1] + s[1]
    return sm, total
sm, count = my_reduce(A, add_sm)

print(f"Sum: {sm}, Count: {count}")
print("Mean: ", sm/count)

# usually in big input the framework can find mean in this way 
```

```python
# using inbuilt methods

A = [1,2,3, 4]

def to_tuple(x):
    return (x , 1)
A = map(to_tuple, A)

print("After map: ", A)
# print("After map: ", list(A))

# You consume the iterator returned by map.
# So by the time you get to:
# A is already empty, and reduce fails.

# can be write like below
# A = list(map(to_tuple, A))


def add_sm(f, s):
    sm = f[0] + s[0]
    total = f[1] + s[1]
    return sm, total

sm, count = reduce(add_sm, A)

print(f"Sum: {sm}, Count: {count}")
print("Mean: ", sm/count)
```
